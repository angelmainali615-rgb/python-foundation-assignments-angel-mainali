# Exercise 3: Clean Numeric Values


# Using loop + continue + isinstance

```python
raw_values = [100, None, 250, "invalid", 300, None, 450]
```

```python
clean_values = []

for value in raw_values:
    # Keep only integers
    if isinstance(value, int):
        clean_values.append(value)
    else:
        continue  # skip non-integers

print(clean_values)  # [100, 250, 300, 450]
```

# Using list comprehension

```python
clean_values=[value for value in raw_values if isinstance(value,int)]
print(clean_values)
```
