# Exercise 7: Nested Order Summary

```python
orders = {
    "ORD-001": {
        "customer": "Anisha",
        "amount": 2500,
        "status": "Completed"
    },
    "ORD-002": {
        "customer": "Ravi",
        "amount": 1800,
        "status": "Pending"
    },
    "ORD-003": {
        "customer": "Maya",
        "amount": 3200,
        "status": "Completed"
    }
}
```

```python
# 1. Print every order ID and customer
for order_id, info in orders.items():
    print(f"Order ID: {order_id}, Customer: {info['customer']}")

# 2. Print only completed orders
print("\nCompleted orders:")
for order_id, info in orders.items():
    if info["status"] == "Completed":
        print(order_id, info)

# 3. Total amount of completed orders
total_completed_amount = 0
for info in orders.values():
    if info["status"] == "Completed":
        total_completed_amount += info["amount"]
print("\nTotal amount of completed orders:", total_completed_amount)

# 4. Count pending orders
pending_count = 0
for info in orders.values():
    if info["status"] == "Pending":
        pending_count += 1
print("Number of pending orders:", pending_count)

# 5. Add a new order
orders["ORD-004"] = {
    "customer": "Sagar",
    "amount": 2100,
    "status": "Pending"
}

print("\nOrders after adding ORD-004:")
for order_id, info in orders.items():
    print(order_id, info)
```
