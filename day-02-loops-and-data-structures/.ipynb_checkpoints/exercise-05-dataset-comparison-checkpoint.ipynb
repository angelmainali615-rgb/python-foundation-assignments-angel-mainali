# Exercise 5: Dataset Comparison

```python
dataset_a = {
    "customer",
    "sales",
    "product",
    "employee"
}

dataset_b = {
    "sales",
    "product",
    "supplier",
    "inventory"
}

# All unique dataset names (union)
all_datasets = dataset_a | dataset_b  # or dataset_a.union(dataset_b)

# Datasets in both groups (intersection)
common_datasets = dataset_a & dataset_b  # or dataset_a.intersection(dataset_b)

# Only in dataset_a (difference)
only_in_a = dataset_a - dataset_b

# Only in dataset_b (difference)
only_in_b = dataset_b - dataset_a

print("All unique dataset names:", all_datasets)
print("Datasets in both groups:", common_datasets)
print("Datasets only in dataset_a:", only_in_a)
print("Datasets only in dataset_b:", only_in_b)
```
