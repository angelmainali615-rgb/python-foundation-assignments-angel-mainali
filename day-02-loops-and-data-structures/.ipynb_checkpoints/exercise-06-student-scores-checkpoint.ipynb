# Exercise 6: Student Score Dictionary

```python
student_scores = {
    "Anisha": 78,
    "Ravi": 55,
    "Maya": 92,
    "Sagar": 61,
    "Nima": 48
}
```

```python
# 1. Print every student and score
for name, score in student_scores.items():
    print(f"{name}: {score}")

# 2. Dictionary with students who scored at least 60 (dict comprehension)
passing_students = {name: score for name, score in student_scores.items() if score >= 60}
print("Passing students:", passing_students)

# 3. Find student with highest score
# max() on items, key by score
top_student, top_score = max(student_scores.items(), key=lambda item: item[1])
print("Top student:", top_student, "with score", top_score)

# 4. Average score
total_score = sum(student_scores.values())
average_score = total_score / len(student_scores)
print("Average score:", average_score)
```
