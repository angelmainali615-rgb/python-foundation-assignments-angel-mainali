# Exercise 2: Retry Simulation

```python
# Exercise 2: Retry Simulation

attempt = 1
max_attempts = 3
operation_successful = False

while attempt <= max_attempts:
    print(f"Attempt {attempt}")
    
    attempt += 1  # go to next attempt

# After the loop
if operation_successful:
    print("Operation completed successfully")
else:
    print("Operation failed after three attempts")
```

```python
# Stretch: simulate success on second attempt

attempt = 1
max_attempts = 3
operation_successful = False

success_on_attempt = 2  # simulate success on attempt 2

while attempt <= max_attempts:
    print(f"Attempt {attempt}")
    
    # Simulate success
    if attempt == success_on_attempt:
        operation_successful = True
        break  # stop trying more attempts
    
    attempt += 1

# After the loop
if operation_successful:
    print("Operation completed successfully")
else:
    print("Operation failed after three attempts")
```
